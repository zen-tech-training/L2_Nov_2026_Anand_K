```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")
print('Model is ready to use!')
```

```python
# Example 1: String output parser
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

strParser = StrOutputParser()

user_prompt_template = PromptTemplate(
    template="Give me 5 lines of blog on Generative AI"
)

chain = user_prompt_template | model | strParser
response = chain.invoke({})

print(type(response))
print(response)
```

```python
# Example 2: List output parser

from langchain_core.output_parsers import ListOutputParser
from langchain_core.prompts import PromptTemplate

class CustomListOutputParser(ListOutputParser):
    def parse(self, text: str) -> list[str]:
        items = [item.strip() for item in text.split(',')]
        return items

listParser = CustomListOutputParser()

user_prompt_template = PromptTemplate(
    template="Please give me the list of comma separated only names of top 5 countries by their economy size"
)

chain = user_prompt_template | model | listParser
response = chain.invoke({})

print(type(response))
print(response)
```

```python
# Example 3: JSON output parser

from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str = Field('Person name')
    age: int = Field('Age of the person')
    city: str = Field('City where person lives')

jsonParser = JsonOutputParser(pydantic_object=Person)

user_prompt_template = PromptTemplate(
    template="Extract the person's details from the following text: John Doe and 30 years old and lives in New York. \nFormat instructions are {format_instructions}",
    partial_variables={'format_instructions': jsonParser.get_format_instructions()}

)

chain = user_prompt_template | model | jsonParser
response = chain.invoke({})

print(type(response))
print(response)
```

```python
# Example 4: Pydantic output parser

from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.prompts import PromptTemplate
from pydantic import BaseModel, Field

class Person(BaseModel):
    name: str = Field('Person name')
    age: int = Field('Age of the person')
    city: str = Field('City where person lives')

pydanticParser = PydanticOutputParser(pydantic_object=Person)

user_prompt_template = PromptTemplate(
    template="Extract the person's details from the following text: John Doe and 30 years old and lives in New York.\nFormat instructions are {format_instructions}",
    partial_variables={'format_instructions': pydanticParser.get_format_instructions()}
)

chain = user_prompt_template | model | pydanticParser
response = chain.invoke({'city': 'Pune'})

print(type(response))
print(response)
```
