```python
# Vector embeddings
from langchain_ollama import OllamaEmbeddings

embedding_model = OllamaEmbeddings(model="nomic-embed-text")
print('Embedding model is ready')
```

```python
user_prompt = "I like music"
words = user_prompt.split()
vector_store = {}

for word in words:
    vector_embedding = embedding_model.embed_query(word)
    print(f'{word} - Vector size {len(vector_embedding)} - {vector_embedding[:10]}')
    vector_store[word] = vector_embedding

print(f'\nFinal vector store: {vector_store}')
```

```python
# Example 2 - FAISS vector store
from langchain_community.vectorstores import FAISS

user_prompt = "I like music"
words = user_prompt.split()
faiss_vector_store = FAISS.from_texts(words, embedding_model)
print(f'FAISS vector store - {faiss_vector_store}')

faiss_vector_store.save_local("faiss_index")

```

```python
vector_store = FAISS.load_local(
    "faiss_index",
    embedding_model,
    allow_dangerous_deserialization=True,
)
print(f'Vector store from FAISS folder {vector_store}')
```
