```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")
print('Model is ready...')
```

```python
from langchain_core.prompts import PromptTemplate

user_prompt_template = PromptTemplate(
    template="Which is the capital of {country}?",
    input_variables=['country']
)

user_prompt = user_prompt_template.format(country='France')
response = model.invoke(user_prompt)
print(response)
```

```python
from langchain_core.prompts import PromptTemplate

user_prompt_template = PromptTemplate(
    template="Please give me 3 lines information about the city {city} located in the country {country}?",
    input_variables=['city', 'country']
)

user_prompt = user_prompt_template.format(country='France', city='Paris')
response = model.invoke(user_prompt)
print(response)
```
