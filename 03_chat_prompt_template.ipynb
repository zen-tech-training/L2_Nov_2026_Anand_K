```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2")
print('Model is ready...')
```

```python
from langchain_core.prompts import ChatPromptTemplate

#user_prompt_template = PromptTemplate(
#    template="Which is the capital of {country}?",
#    input_variables=['country']
#)

user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Which is the capital of {country}?")
])
user_chat_prompt = user_chat_prompt_template.invoke({"country": "France"})

response = model.invoke(user_chat_prompt)
print(response)
```

```python
# Roles in ChatPromptTemplate
from langchain_core.prompts import ChatPromptTemplate

user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Explain the {theorem} in 3 lines"),
    ("system", "You are an expert mathematician who has knowledge about all mathematics theorems")
])
user_chat_prompt = user_chat_prompt_template.invoke({"theorem": "pythagoras theorem"})

response = model.invoke(user_chat_prompt)
print(response)
```
