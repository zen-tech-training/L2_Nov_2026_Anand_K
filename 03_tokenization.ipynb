```python
import tiktoken

tiktoken = tiktoken.get_encoding("cl100k_base")

#user_prompt = "Which is the capital city of France in the european continent?"
user_prompt = "What is a Generative AI?"

input_tokens = tiktoken.encode(user_prompt)
print(f'Number of tokens {len(input_tokens)}')

for input_token in input_tokens:
    print(f'{tiktoken.decode([input_token])} - {input_token}')
```
